## 4.1 Importing Statistical Libraries

The `scipy.stats` module is imported to carry out the independent t-test and the accompanying Levene's test for equality of variances.

```python
from scipy import stats
```

## 4.2 Formulating the Hypotheses

Milestone 3 visually suggested that discounting is associated with lower profit, and that profitability differs across product categories. These observations are now tested statistically using the **independent t-test**, which compares the means of two independent groups.

**Test 1 – Discount and Profit**
- H0: There is no difference in mean Profit between discounted and non-discounted orders (μ1 = μ2)
- H1: There is a significant difference in mean Profit between discounted and non-discounted orders (μ1 ≠ μ2)

**Test 2 – Profit Margin: Technology vs Office Supplies**
- H0: There is no difference in mean Profit Margin (%) between the Technology and Office Supplies categories (μ1 = μ2)
- H1: There is a significant difference in mean Profit Margin (%) between the Technology and Office Supplies categories (μ1 ≠ μ2)

**Test 3 – Sales: Central vs East Region**
- H0: There is no difference in mean Sales between the Central and East regions (μ1 = μ2)
- H1: There is a significant difference in mean Sales between the Central and East regions (μ1 ≠ μ2)

For all tests, a significance level of **α = 0.05** is used.

## 4.3 Test 1: Effect of Discount on Profit

Orders are split into two independent groups — discounted (`Discount` > 0) and non-discounted (`Discount` = 0) — and their mean Profit is compared.

```python
discounted = df[df["Discount"] > 0]["Profit"]
no_discount = df[df["Discount"] == 0]["Profit"]

print("Discounted orders   -> n:", len(discounted), " mean profit:", round(discounted.mean(), 2))
print("Non-discounted orders -> n:", len(no_discount), " mean profit:", round(no_discount.mean(), 2))

levene_stat, levene_p = stats.levene(discounted, no_discount)
print(f"\nLevene's test statistic: {levene_stat:.4f}, p-value: {levene_p:.4f}")

equal_var = levene_p > 0.05

t_stat, p_value = stats.ttest_ind(discounted, no_discount, equal_var=equal_var)

print(f"\nT-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.4f}")

alpha = 0.05
if p_value < alpha:
    print("\nResult: Reject H0 - there is a statistically significant difference in mean Profit between discounted and non-discounted orders.")
else:
    print("\nResult: Fail to reject H0 - no statistically significant difference in mean Profit was found.")
```

### Observation

The Levene's test result indicates whether the equal-variance assumption holds, which determines whether the standard Student's t-test or Welch's t-test is used. The resulting p-value is compared against α = 0.05 to determine whether discounting has a statistically significant effect on Profit, confirming (or challenging) the pattern seen in the Milestone 3 scatter plot of Discount vs Profit.

## 4.4 Test 2: Profit Margin — Technology vs Office Supplies

The two highest-volume product categories are compared to statistically validate the differences in average Profit Margin (%) observed in Milestone 3.

```python
tech_margin = df[df["Category"] == "Technology"]["Profit Margin (%)"]
office_margin = df[df["Category"] == "Office Supplies"]["Profit Margin (%)"]

print("Technology       -> n:", len(tech_margin), " mean margin:", round(tech_margin.mean(), 2))
print("Office Supplies  -> n:", len(office_margin), " mean margin:", round(office_margin.mean(), 2))

levene_stat2, levene_p2 = stats.levene(tech_margin, office_margin)
print(f"\nLevene's test statistic: {levene_stat2:.4f}, p-value: {levene_p2:.4f}")

equal_var2 = levene_p2 > 0.05

t_stat2, p_value2 = stats.ttest_ind(tech_margin, office_margin, equal_var=equal_var2)

print(f"\nT-statistic: {t_stat2:.4f}")
print(f"P-value: {p_value2:.4f}")

if p_value2 < alpha:
    print("\nResult: Reject H0 - Profit Margin (%) differs significantly between Technology and Office Supplies.")
else:
    print("\nResult: Fail to reject H0 - no statistically significant difference in Profit Margin (%) was found.")
```

### Observation

This test statistically confirms whether the difference in average Profit Margin (%) between Technology and Office Supplies seen in the Milestone 3 bar chart (Section 3.2) is a genuine effect or could plausibly be due to random variation in the sample.

## 4.5 Test 3: Sales — Central vs East Region

Sales are compared between the Central and East regions to determine whether the regional differences observed in Milestone 1 (Section 1.8.2) are statistically significant.

```python
central_sales = df[df["Region"] == "Central"]["Sales"]
east_sales = df[df["Region"] == "East"]["Sales"]

print("Central -> n:", len(central_sales), " mean sales:", round(central_sales.mean(), 2))
print("East    -> n:", len(east_sales), " mean sales:", round(east_sales.mean(), 2))

levene_stat3, levene_p3 = stats.levene(central_sales, east_sales)
print(f"\nLevene's test statistic: {levene_stat3:.4f}, p-value: {levene_p3:.4f}")

equal_var3 = levene_p3 > 0.05

t_stat3, p_value3 = stats.ttest_ind(central_sales, east_sales, equal_var=equal_var3)

print(f"\nT-statistic: {t_stat3:.4f}")
print(f"P-value: {p_value3:.4f}")

if p_value3 < alpha:
    print("\nResult: Reject H0 - mean Sales differs significantly between the Central and East regions.")
else:
    print("\nResult: Fail to reject H0 - no statistically significant difference in mean Sales was found.")
```

### Observation

The result indicates whether the apparent gap in total sales between the Central and East regions reflects a genuine difference in average order value, or whether it is more likely driven by differences in the number of orders placed rather than the size of individual transactions.

## 4.7 Summary of Hypothesis Testing

The independent t-tests move the analysis from visual, descriptive observations (Milestone 3) to inferential conclusions: they quantify how confident we can be that the differences seen in the charts reflect real underlying differences in the population of orders, rather than random sampling variation. Where a test is significant, the corresponding Cohen's d indicates whether the effect is large enough to be practically meaningful for business decision-making (e.g. discount policy, category investment, or regional strategy).

## 4.8 Chi-Square Test of Independence

The t-tests above compare means of a numeric variable across two groups. To test whether two **categorical** variables are associated (rather than independent), a chi-square test of independence is used. Here we test whether **Ship Mode** is associated with **Segment**.

- H0: Ship Mode and Segment are independent (no association)
- H1: Ship Mode and Segment are associated

```python
contingency_table = pd.crosstab(df["Ship Mode"], df["Segment"])
contingency_table
```

```python
chi2_stat, chi2_p, dof, expected = stats.chi2_contingency(contingency_table)

print(f"Chi-square statistic: {chi2_stat:.4f}")
print(f"Degrees of freedom:   {dof}")
print(f"P-value:              {chi2_p:.4f}")

if chi2_p < alpha:
    print("\nResult: Reject H0 - Ship Mode and Segment are significantly associated.")
else:
    print("\nResult: Fail to reject H0 - no significant association found between Ship Mode and Segment.")
```

### Observation

The chi-square test evaluates whether the distribution of shipping preferences differs meaningfully across customer segments. A significant result would suggest that segment-specific shipping strategies (e.g. promoting faster shipping to a particular segment) could be worthwhile; a non-significant result suggests shipping preference is unrelated to segment.

## 4.9 Correlation Analysis

The Milestone 3 heatmap (Section 3.6) showed the *strength* of linear relationships between numeric variables. Here, Pearson correlation coefficients are calculated together with their p-values to test whether each correlation is **statistically significant**, not just descriptively present.

- H0: There is no linear correlation between the two variables (ρ = 0)
- H1: There is a linear correlation between the two variables (ρ ≠ 0)

```python
pairs = [
    ("Sales", "Profit"),
    ("Discount", "Profit"),
    ("Quantity", "Profit"),
    ("Sales", "Profit Margin (%)"),
]

print(f"{'Variable 1':<20}{'Variable 2':<20}{'r':>8}{'p-value':>12}{'Significant?':>15}")
for var1, var2 in pairs:
    r, p = stats.pearsonr(df[var1], df[var2])
    sig = "Yes" if p < alpha else "No"
    print(f"{var1:<20}{var2:<20}{r:>8.3f}{p:>12.4f}{sig:>15}")
```

### Observation

Statistically significant correlations (p < 0.05) confirm that the relationships observed visually in Milestone 3 are unlikely to be due to chance. The strength (r) indicates how strong each relationship is: values near 0 indicate a weak linear relationship, while values closer to ±1 indicate a strong one. Note that a statistically significant correlation does not by itself imply causation — for example, a significant negative correlation between Discount and Profit indicates association, not proof that discounts directly cause profit loss.

## 4.10 Regression Analysis

A simple linear regression is used to model **Profit as a function of Discount**, quantifying the relationship and testing whether Discount is a statistically significant predictor of Profit.

```python
x = df["Discount"]
y = df["Profit"]

slope, intercept, r_value, p_value_reg, std_err = stats.linregress(x, y)

print(f"Regression equation: Profit = {intercept:.3f} + ({slope:.3f}) * Discount")
print(f"R-squared:           {r_value**2:.4f}")
print(f"P-value (slope):     {p_value_reg:.6f}")
print(f"Standard error:      {std_err:.4f}")

if p_value_reg < alpha:
    print("\nResult: Discount is a statistically significant predictor of Profit.")
else:
    print("\nResult: Discount is not a statistically significant predictor of Profit.")
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(x, y, alpha=0.4, label="Observed data")
x_line = np.linspace(x.min(), x.max(), 100)
y_line = intercept + slope * x_line
plt.plot(x_line, y_line, color="red", label="Fitted regression line")
plt.title("Linear Regression: Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.legend()
plt.show()
```

### Observation

The R-squared value indicates the proportion of variability in Profit explained by Discount alone; a low R-squared suggests other factors (Category, Quantity, Region, etc.) also drive profitability. The p-value for the slope tests whether the relationship is statistically distinguishable from zero, providing a formal, model-based counterpart to the Milestone 1 scatter plot of Discount vs Profit.

## 4.11 Time Series / Trend Analysis

The Milestone 3 line chart (Section 3.1) showed monthly sales fluctuations. Here, a linear trend is fitted over time to test whether sales show a statistically significant upward or downward trend across the year.

```python
month_order = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
]

monthly_sales_trend = df.groupby("Order Month")["Sales"].sum().reindex(month_order)

time_index = np.arange(len(monthly_sales_trend))
sales_values = monthly_sales_trend.values

slope_t, intercept_t, r_value_t, p_value_t, std_err_t = stats.linregress(time_index, sales_values)

print(f"Trend equation: Sales = {intercept_t:.2f} + ({slope_t:.2f}) * Month_Index")
print(f"R-squared:      {r_value_t**2:.4f}")
print(f"P-value:        {p_value_t:.4f}")

if p_value_t < alpha:
    direction = "increasing" if slope_t > 0 else "decreasing"
    print(f"\nResult: Statistically significant {direction} trend in monthly sales.")
else:
    print("\nResult: No statistically significant trend in monthly sales - fluctuations appear to be random/seasonal rather than a consistent trend.")
```

```python
plt.figure(figsize=(10, 5))
plt.plot(month_order, sales_values, marker="o", label="Monthly Sales")
plt.plot(month_order, intercept_t + slope_t * time_index, color="red", linestyle="--", label="Trend line")
plt.title("Monthly Sales Trend with Fitted Trend Line")
plt.xlabel("Order Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.show()
```

### Observation

Fitting a trend line statistically validates whether the fluctuations seen in Milestone 3 reflect a genuine directional trend over the year, or whether sales are essentially flat with random/seasonal variation around a stable average. This distinction matters for forecasting and planning purposes.

## 4.12 Interpretation of Uncertainty and Variability

Point estimates (means, totals) do not convey how much uncertainty surrounds them. 95% confidence intervals are calculated for key metrics to express the range within which the true population mean is likely to fall.

```python
def confidence_interval(data, confidence=0.95):
    data = np.array(data)
    n = len(data)
    mean = data.mean()
    sem = stats.sem(data)  # standard error of the mean
    margin = sem * stats.t.ppf((1 + confidence) / 2, n - 1)
    return mean, mean - margin, mean + margin

metrics = {
    "Overall Profit": df["Profit"],
    "Overall Sales": df["Sales"],
    "Discounted Orders - Profit": discounted,
    "Non-discounted Orders - Profit": no_discount,
}

print(f"{'Metric':<32}{'Mean':>12}{'95% CI Lower':>16}{'95% CI Upper':>16}")
for name, series in metrics.items():
    mean, lower, upper = confidence_interval(series)
    print(f"{name:<32}{mean:>12.2f}{lower:>16.2f}{upper:>16.2f}")
```

### Observation

The confidence intervals show the precision of each estimate. Narrow intervals indicate more reliable estimates (typically from larger or less variable groups), while wide intervals signal greater uncertainty. Where the confidence intervals for two groups (e.g. discounted vs non-discounted Profit) do not overlap, this reinforces the conclusion of a significant difference found in the t-test; overlapping intervals suggest more caution is warranted even if a test returned a low p-value.
